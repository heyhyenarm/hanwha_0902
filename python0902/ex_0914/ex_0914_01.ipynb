```python
from langchain_openai import ChatOpenAI
```

```python
# 챗 모델 객체 생성
llm = ChatOpenAI(
    temperature=0.1,    # 창의성
    model= "gpt-5.6-luna",  # 모델명
)

# 질문
question = "일본의 수도는 어디인가요?"

# 응답
response = llm.invoke(question)
```

```python
print(f"[답변]: {response}")
```

```python
response.content
```

```python
response.response_metadata
```

```python
# 사용된 전체 토큰: 24
response.response_metadata["token_usage"]["total_tokens"]
```

```python
# 답변을 한 토큰씩 받아와 화면에 실시간으로 출력
answer = llm.stream("에버랜드역 도보 20분 이내 맛집 10곳과 대표 메뉴. 에버랜드 내부X")
```

```python
# flush=False: 결과가 버퍼에 저장되었다가 한번에 출력
# flush=True: 결과가 생성되는대로 바로 출력
for token in answer:
    print(token.content, end="", flush=True)
```

```python
# 스트리밍 답변 결과 변수에 저장
answer = llm.stream("에버랜드 내부 식당과 대표 메뉴는?")

final_answer = ""
for token in answer:
    final_answer += token.content
```

```python
# 저장된 결과 출력
final_answer
```
